```python
def describe_passenger(p):
    print(f"{p['Name']} | пол: {p['Sex']} | возраст: {p['Age']} | класс: {p['Pclass']}")

passenger = {"Name": "Braund, Mr. Owen Harris", "Age": 22,
             "Sex": "male", "Pclass": 3, "Survived": 0}
describe_passenger(passenger)
```

```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

print(type(df))
print("Загружено")
```

```python
display(df.head())

print("Размер:", df.shape)

print("Колонки:", list(df.columns))
```

```python
df.info()
```

```python
display(df[["Age", "Fare", "SibSp", "Parch"]].describe())
```

```python
print("Survived:")
print(df["Survived"].value_counts())

print("\nSex:")
print(df["Sex"].value_counts())

print("\nPclass:")
print(df["Pclass"].value_counts())

print("\nSurvived (доли):")
print(df["Survived"].value_counts(normalize=True))
```

```python
survival_by_sex = df.groupby("Sex")["Survived"].mean()
print(survival_by_sex)

assert survival_by_sex["female"] > survival_by_sex["male"], "Ожидание не сбылось"
print("Инсайт подтверждён: женщин спасали чаще.")
```
